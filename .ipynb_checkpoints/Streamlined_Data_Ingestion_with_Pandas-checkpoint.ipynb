## Loading select sheets
---

```python
# Get the second sheet by position index
survey_data_sheet2 = pd.read_excel("fcc_survey.xlss",
                                   sheet_name = 1) # Base-0 indexed

# Get the second sheet by name
survey_data_2017 = pd.read_excel("fcc_survey.xlsx",
                                sheet_name = '2017')

# Get all sheets
survey_responses = pd.read_excel("fcc_survey.xlsx",
                                 sheet_name = None)
print(type(survey_responses)) # dictionary
```

## Combining all sheets
---

```python
# Create empty dataframe
all_responses = pd.DataFrame()

# Iterate through dataframes in dictionary
for sheet_name, df in survey_responses.items():
    # Add year columns
    df["year"] = sheet_name

    # Add the df to all_responses
    all_responses = pd.concat([all_responses, df])

# View years in data
print(all_responses["year"].unique())
```

## Parsing Dates
---

```python
# List columns of dates to parse
data_cols = ["Part1StartTime", "Part1EndTime"]

# Load file, parsing standard datetime columns
surver_df = pd.read_excel("fcc_survery.xlsx",
                          parse_dates = data_cols)
```

## Combining multiple dates columns

```python
# List columns of dates to parse
date_cols = ["Part1StartTime",
             "Part1EndTime",
             [["Part2StartDate", "Part2StartTime"]] # Combine
             ]

survey_df = pd.read_excel("fcc_survey.xlsx",
                          parse_dates = data_cols)
print(survey_df.head(3))


## Change name
# List columns of dates to parse
date_cols = {"Part1Start": "Part1StartTime",
             "Part1End": "Part1EndTime",
             "Part2Start": [["Part2StartDate", "Part2StartTime"]] # Combine
             }

survey_df = pd.read_excel("fcc_survey.xlsx",
                          parse_dates = data_cols)
print(survey_df.head(3))
```

## Parsing Non-Standard Dates

```python
format_string = "%m%d%Y %H:%M:%S"
survey_df["Part2EndTime"] = pd.to_datetiem(survey_df["Part2EndTime"],
                                           format = format_string)
```
